### Example 1 (Margin of error) 
A manufacturing engineer wants to use the mean of a random sample of size n = 64 to estimate the mean length of the rods being manufactured. If it is known that σ = 0.5 cm, then find the margin of error with 95% probability.

```python
# E = Z_alpha/2 * sigma / sqrt(n)
n = 64
sigma = 0.5
alpha = 0.05
z_alpha = norm.ppf(1-alpha/2)
E = z_alpha * sigma / sqrt(n)
E
```

## When the Population is Finite
If the population is finite, the standard error is $\frac{σ}{\sqrt{n}} \sqrt{\frac{N-n}{N-1}}$ where N is the population size and $\sqrt{\frac{N-n}{N-1}}$ is called the **finite population correction factor**. An analysis reveals that if the population size is large relative to the sample size, then the finite population correction factor is close to 1 and can be ignored. <br>

**The correction factor is ignored if either the sampling is done with replacement or  or the population size is at least 20 times larger than the sample size.**

In practice, most applications involve populations that qualify as large because if the population is small, it may be possible to investigate each member of the population, and in so doing, calculate the parameters precisely. As a consequence, the finite population correction factor is usually omitted.

### Example 5 (Finding one-sided confidence intervals)
A manufacturing engineer decided to check the efficiency of a new technician hired by the company. She records the time taken by the technician to complete 100 randomly selected jobs and found that in this sample of 100, the average time taken per job was 10 hours with a standard deviation of two hours. Find 95% lower and upper one-sided confidence intervals for μ.

```python
n = 100
x_bar = 10
s = 2
significance = 0.95

# The sample size is large, so I use Z instead of t
z_alpha = norm.ppf(significance)
l = x_bar - z_alpha * s / sqrt(n)
u = x_bar + z_alpha * s / sqrt(n)
l,u
```

**Interpretation:** We are 95% confident that the newly hired technician would take a maximum average time of 10.329 hours and a minimum average time of 9.671 hours to complete the job.

### Example 5: Does the mean weekly newspaper collection exceeds 2.0 pounds?
In the near future, nations will likely have to do more to save the environment. Possible
actions include reducing energy use and recycling. Currently, most products manufactured from recycled material are considerably more expensive than those manufactured
from material found in the earth. For example, it is approximately three times as expensive to produce glass bottles from recycled glass than from silica sand, soda ash, and limestone, all plentiful materials mined in numerous countries. It is more expensive to manufacture aluminum cans from recycled cans than from bauxite. Newspapers are an
exception. It can be profitable to recycle newspaper. A major expense is the collection
from homes.
<br>In recent years, many companies have gone into the business of collecting
used newspapers from households and recycling them. A financial analyst for one such
company has recently computed that the firm would make a profit if the mean weekly
newspaper collection from each household exceeded 2.0 pounds.
<br>In a study to determine the feasibility of a recycling plant, a random sample of 148 households was drawn
from a large community, and the weekly weight of newspapers discarded for recycling
for each household was recorded and listed. Do these data provide sufficient evidence to allow the analyst to conclude that a recycling plant would be profitable?

**Solution**
<br>The problem objective is to describe the population of the amounts of newspaper discarded by each household in the population. The data are interval, indicating that the
parameter to be tested is the population mean. Because the financial analyst needs to
determine whether the mean is greater than 2.0 pounds, the alternative hypothesis is:<br>
$ H_1: μ > 2.0 $
<br>As usual, the null hypothesis states that the mean is equal to the value listed in the alternative hypothesis:<br>
$ H_0: μ = 2.0 $

```python
df = pd.read_csv("Data\\recyling newspaper.csv", index_col='i')
df.shape
```

The manager believes that the cost of a Type I error (concluding that the mean is greater
than 2 when it isn’t) is quite high. Consequently, he sets the significance level at 99%.

```python
significance = 0.99
mu = 2

n = df.shape[0]
x_bar = np.mean(df)
std = np.std(df, ddof=1).iloc[0]
threshold = t.ppf(significance, df = n-1)
t_stat = (x_bar - mu) / (std/sqrt(n))

print(f"t = {threshold}")
print(f"t statistics = {t_stat}")
if t_stat > threshold:
    print("The null hypothesis is rejected in favor of alternative hypothesis. So the population mean is greater than 2.")
else:
    print("We reject the alternative hypothesis in favor of null hypothesis. Consequently, the population mean in not greater than 2.")
```

```python
p_value = 1 - t.cdf(threshold, df=n-1)
p_value
```

The p_value < 0.025 so the null hypothesis is rejceted.

### Example 6: Tax Collected from Audited Returns
In 2014 (the latest year reported), 146,861,217 tax returns were filed in the United States. The Internal Revenue Service (IRS) examined 1,228,117, of them to determine if they were correctly done.<br>
To determine how well the auditors are performing, a random sample of these returns was drawn and the additional tax was reported, which is listed in the tax.csv file. Estimate with 95% confidence the mean additional income tax collected from the 1,228,117 files audited.<br>
**Solution:** The problem objective is to describe the population of additional income tax. The data are interval and hence, the parameter is the population mean μ. The question asks us to estimate this parameter.

```python
df = pd.read_csv("Data\\tax.csv", index_col='i')
df.shape
```

```python
significance = 0.95
sl= (1 - significance) / 2
sr = 1 - sl

n = df.shape[0]
x_bar = np.mean(df)
std = np.std(df, ddof=1, axis=0).iloc[0]
tl, tr = t.ppf([sl,sr], df = n-1)
LCL = x_bar + tl * (std/sqrt(n))
UCL = x_bar + tr * (std/sqrt(n))

LCL, UCL
```

We estimate that the mean additional tax collected lies between 8, 887 and 10, 168 dolars.
<br>The 95% confidence interval estimate of the total is:

```python
population = 1228117
PLCL, PUCL = (population * LCL, population * UCL)
PLCL, PUCL
```
