# تمرین 1
برنامه ای بنویسید که دو عدد از ورودی بگیرد و جمع، تفریق، ضرب و تقسیم آنها را حساب کند.

# تمرین 2
برنامه ای بنویسید که یک لیست از اسم خوراکی ایجاد کند و طول لیست را چاپ کند.

سپس چند خوراکی جدید به لیست اضافه کند و دوباره طول لیست را چاپ کند.

سپس چند خوراکی را از لیست حذف کند و طول لیست را چاپ کند.

# تمرین 3
