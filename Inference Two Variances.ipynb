## Example 1 (Confidence intervals for the ratio of two population variances) 
Two random samples of sizes 13 and 16 are selected from a group of patients with hypertension.
The patients in the two samples are independently treated with drugs A and B. After a full course of treatments, these patients are evaluated. The data collected at the time of evaluation yielded sample standard deviations SA = S1 = 6.5 mmHg and SB = S2 = 7.5 mmHg. Assuming that the two sets of data come from independent normal populations with variances $σ_1^2$ and $σ_2^2$, respectively, determine a 95% two-sided confidence interval for $\frac{σ_1^2}{σ_2^2}$. Also determine the one-sided confidence intervals for $\frac{σ_1^2}{σ_2^2}$ and σ1/σ2.

```python
# 95% confidence interval for variance1/variance2
n1 = 13
n2 = 16
s1 = 6.5
s2 = 7.5
significance = 0.95
alpha = 1 - significance
alpha1 = alpha / 2
alpha2 = 1 - alpha1

fl = f.ppf(alpha1, n1-1, n2-1)
fu = f.ppf(alpha2, n1-1, n2-1)
# fl , fu
l = s1**2 / (s2**2 * fu)
u = s1**2 / (s2**2 * fl)
print(f"The 95% confidence interval for variance1/variance2 is: ({l}, {u})")
```

```python
# 95% confidence interval for σ1 / σ2
print(f"The 95% confidence interval for variance1/variance2 is: ({sqrt(l)}, {sqrt(u)})")
```

**Interpretation:** The two-sided confidence intervals contain 1, which means that at the 5% level of significance, we can conclude that two variances are not significantly different from each other.

```python
# The 95% lower and upper one-sided confidence intervals for variance1/variance2
n1 = 13
n2 = 16
s1 = 6.5
s2 = 7.5
alpha = 0.95

fl = f.ppf(alpha, n1-1, n2-1)
fu = f.ppf(1-alpha, n1-1, n2-1)
# fu = f.ppf(0.95, 15,12)

# fl , fu
l = s1**2 / (s2**2 * fu)
u = s1**2 / (s2**2 * fl)
print(f"One sided 95% confidence interval for variance1/variance2: (0, {l.round(4)})")
print(f"One sided 95% confidence interval for variance1/variance2: ({u.round(4)}, +∞)")
```

```python
l = s1 / (s2 * sqrt(fu))
u = s1 / (s2 * sqrt(fl))
print(f"One sided 95% confidence interval for σ2: (0, {round(l, 4)})")
print(f"One sided 95% confidence interval for σ2: ({round(u, 4)}, +∞)")
```
